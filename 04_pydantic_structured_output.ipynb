```python
from pydantic import BaseModel, Field
from typing import Optional, Literal
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv 
```

```python
load_dotenv()
```

```python
class Review(BaseModel):
    Product : str = 'Not Available'
    Pros : list[str]
    Cons : list[str]
    Reviewer: Optional[str] = None
    ratings:float = Field(gt=0, lt=5, default=1, description='Return rating of the reviewer for the Current Product')
```

```python
class Data(BaseModel):
    Product: str = Field(description='Provide the Name of the Product')
    Pros: list[str] = Field(description='Mention positive points mentioned about the Product')
    Cons: list[str] = Field(description='Mention Negative points mentioned about the Product')
    Sentiment: Literal["Positive", 'Negative'] = Field(description='Mention the sentiemnt of the user review')
    Ratings: float = Field(ge=1, le=5, default=1, description='Rating provided by user as per the Review')
    Reviewer : Optional[str] = Field(description='Name of the reviewer', default=None)
```

```python
Review = '''I recently bought the Sony WH-1000XM5 headphones and I'm very impressed. The noise cancellation is excellent, the sound quality is clear and detailed, and the headphones are comfortable even after several hours. The only downside is that they are quite expensive. The touch controls can also be a little sensitive. I'd give them 4.5 out of 5.'''
```

```python
model = ChatOpenAI(model='gpt-3.5-turbo')
```

```python
structured_model = model.with_structured_output(Data)
```

```python
response = structured_model.invoke(Review)
```

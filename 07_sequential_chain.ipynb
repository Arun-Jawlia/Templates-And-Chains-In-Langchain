```python
from langchain_core.prompts import PromptTemplate
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_core.output_parsers import StrOutputParser
from langchain_google_genai import ChatGoogleGenerativeAI
```

# Simple Sequential Chain

```python
prompt = PromptTemplate(template='Create a 5 point summary on {Topic}', input_variable=['Topic'])
```

```python
model = ChatOpenAI(model='gpt-3.5-turbo')
```

```python
parser = StrOutputParser()
```

```python
chain = prompt | model | parser
```

```python
result = chain.invoke({'Topic':"FIFA Cup"})
```

```python
result
```

# Multiple Chains

```python
prompt_1 = PromptTemplate(template='Create a detailed report on {Topic}', input_variable=['Topic'])
```

```python
model_1 = ChatGoogleGenerativeAI(model='gemini-3.7-flash')
```

```python
prompt_2= PromptTemplate(template='Create a 5 point summary on the {Report}', input_variables=['Report'])
```

```python
model_2 = ChatOpenAI(model='gpt-3.5-turbo')
```

```python
parser = StrOutputParser()
```

```python
chain = prompt_1 | model_1 | parser | prompt_2 | model_2 | parser
```

```python
result = chain.invoke({'Topic': 'Cricket World Cup'})
```

```python
result
```

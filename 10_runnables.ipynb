# Runnable Sequence

```python
from langchain_core.runnables import RunnableSequence
```

```python
from langchain_core.prompts import PromptTemplate
from langchain_openai import ChatOpenAI
from langchain_groq import ChatGroq
from langchain_core.output_parsers import StrOutputParser
from dotenv import load_dotenv
import os
```

```python
load_dotenv()
```

```python
llm = ChatGroq(model='llama-3.3-70b-versatile', api_key= os.getenv("GROQ_API_KEY"))
```

```python
prompt = PromptTemplate(template='Generate a Joke on this {Topic}', input_variables=['Topic'])
```

```python
parser = StrOutputParser()
```

```python
chain = RunnableSequence(prompt, llm , parser)
```

```python
result = chain.invoke({"Topic": 'AI can Replace Humans'})
```

# Runnable Passthrought

```python
from langchain_core.runnables import RunnablePassthrough, RunnableParallel, RunnableSequence
```

```python
from langchain_core.prompts import PromptTemplate
from langchain_openai import ChatOpenAI
from langchain_groq import ChatGroq
from langchain_core.output_parsers import StrOutputParser
from dotenv import load_dotenv
import os
```

```python
load_dotenv()
```

```python
prompt = PromptTemplate(template='Generate the joke on {Topic}', input_variables=['Topic'])
```

```python
llm = ChatOpenAI(model='gpt-3.5-turbo', temperature=0.5)
```

```python
parser = StrOutputParser()
```

```python
chain1 = RunnableSequence(prompt, llm , parser)
```

```python
prompt2 = PromptTemplate(template='Generate the Explanation for the following {Joke}', input_variables=['Joke'])
```

```python
explain = prompt | llm | parser | prompt2 | llm | parser
```

```python
response = explain.invoke({'Topic': 'AI vs Humans'})
```

```python
response
```

```python
parallel_chain = RunnableParallel({"Joke": RunnablePassthrough() , 'Explanation' :prompt2 | llm | parser })
```

```python
final_chain = chain1 | parallel_chain
```

```python
final_chain.invoke({'Topic': 'AI vs Humans'})
```

# Runnable Lambda

It is goint to act middleware between different AI Components enabling Preprocessing, Transformation, API Calls etc

```python
from langchain_core.runnables import RunnablePassthrough, RunnableParallel, RunnableSequence, RunnableLambda
```

```python
from langchain_core.prompts import PromptTemplate
from langchain_openai import ChatOpenAI
from langchain_groq import ChatGroq
from langchain_core.output_parsers import StrOutputParser
from dotenv import load_dotenv
import os
```

```python
load_dotenv()
```

```python
def wordN(words):
    return len(words.split())
```

```python
prompt = PromptTemplate(template='Generate the joke on {Topic}', input_variables=['Topic'])
```

```python
llm = ChatOpenAI(model='gpt-3.5-turbo', temperature=0.5)
```

```python
parser = StrOutputParser()
```

```python
chainJ = prompt | llm | parser
```

```python
parallel_chain = RunnableParallel({'joke':RunnablePassthrough(), 'Words Count': RunnableLambda(wordN)})
```

```python
final_chain = chainJ | parallel_chain
```

```python
final_chain.invoke({'Topic':"AI vs Human"})
```

```python
google_model = ChatGoogleGenerativeAI(model='gemini-3.6-flash')
```

```python
gpt_model = ChatOpenAI(model='gpt-3.5-turbo')
```

```python
prompt1 = PromptTemplate(template='Generate a Detailed report on the {Topic}', input_variables=['Topic'])
```

```python
prompt2 = PromptTemplate(template='Create a 5 point summary on {Report}', input_variables=['Report'])
```

```python
final1 = prompt1.invoke({'Topic':"Google vs Openai Models"})
```

```python
parser = StrOutputParser()
```

```python
chain = prompt1 | google_model | parser | prompt2 | gpt_model | parser
```

```python
response = chain.invoke({'Topic':"Google vs Openai Models"})
```
